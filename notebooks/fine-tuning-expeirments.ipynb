# Fine-tuning Training Curves

Visualises accuracy and F1-score vs. epoch for each fine-tuned model.  
Configure experiment paths in the **Configuration** cell below.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

ROOT = Path("..").resolve()

# ── Configuration ──────────────────────────────────────────────────────────────
# Map display name → path to training_logs.csv
EXPERIMENTS = {
    "DistilBERT":           ROOT / "experiments" / "experiment1" / "logs" / "training_logs.csv",
    "BERT-Large":           ROOT / "experiments" / "experiment2" / "logs" / "training_logs.csv",
    "Twitter-RoBERTa-base": ROOT / "experiments" / "experiment4" / "logs" / "training_logs.csv",
}
METRICS = ["train_accuracy", "train_f1_score", "val_accuracy", "val_f1_score"]
REPORTS_DIR = ROOT / "reports"
# ──────────────────────────────────────────────────────────────────────────────

print("Configured experiments:")
for name, path in EXPERIMENTS.items():
    status = "✓" if path.exists() else "✗ (not found)"
    print(f"  {status}  {name}: {path}")
```

## Training curves

```python
for model_name, log_path in EXPERIMENTS.items():
    if not log_path.exists():
        print(f"Skipping {model_name}: {log_path} not found")
        continue

    df = pd.read_csv(log_path).set_index("epoch")
    fig, ax = plt.subplots(figsize=(10, 6))
    df[METRICS].plot(ax=ax)
    ax.set_title(f"{model_name} Fine-tuning — Accuracy / F1-score vs Epochs")
    ax.set_xlabel("Epochs")
    ax.grid(True, linestyle="--", alpha=0.5)
    slug = model_name.lower().replace(" ", "-").replace("/", "-")
    fig.savefig(REPORTS_DIR / f"{slug}-performance.png", bbox_inches="tight", dpi=150)
    plt.show()
    plt.close(fig)
```
