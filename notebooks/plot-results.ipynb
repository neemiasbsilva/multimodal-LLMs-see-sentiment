# Results Visualization

Loads all F1-score data from `reports/results.json` and renders comparison plots
using helpers from `utils/visualization.py`.  
**No hardcoded numbers live in this notebook** — update `results.json` to add new results.

```python
import json
import sys
from pathlib import Path
import matplotlib.pyplot as plt
%matplotlib inline

ROOT = Path("..").resolve()
sys.path.insert(0, str(ROOT))

from utils.visualization import (
    scatter_by_model,
    scatter_grid,
    comprehensive_errorbar,
    confidence_band_plot,
    bar_comparison,
)

RESULTS_PATH = ROOT / "reports" / "results.json"
with open(RESULTS_PATH) as f:
    results = json.load(f)

print(f"Loaded {len(results['comprehensive']['records'])} records from {RESULTS_PATH.name}")
print("Approaches:", list(results['comprehensive']['approach_styles'].keys()))
```

## Scatter plots — σ=5, one per classification problem

```python
fig = scatter_grid(results, figsize=(18, 14))
fig.savefig(ROOT / "reports" / "scatter-sigma5-all-problems.png", bbox_inches="tight", dpi=150)
plt.show()
```

## Comprehensive errorbar — all approaches × all thresholds

```python
fig, ax = plt.subplots(figsize=(20, 8))
comprehensive_errorbar(results, ax=ax)
plt.tight_layout()
fig.savefig(ROOT / "reports" / "f1-score-comparison-errorbar.png", bbox_inches="tight", dpi=150)
plt.show()
```

## Confidence-band plot — best approaches at σ=5

```python
best_approaches = [
    "ResNet - (Benchmark)",
    "MiniGPT-4 (OpenAI) + Fine-Tuning DISTIL-BERT",
]

fig, ax = plt.subplots(figsize=(20, 8))
confidence_band_plot(results, selected_approaches=best_approaches, threshold=5, ax=ax)
plt.tight_layout()
plt.show()
```

## Bar chart — ResNet vs best fine-tuned approaches across all thresholds

```python
comparison_approaches = [
    "ResNet - (Benchmark)",
    "MiniGPT-4 (OpenSource) + Fine-Tuning BART-LARGE-MNLI",
    "MiniGPT-4 (OpenAI) + Fine-Tuning DISTIL-BERT",
]

fig, ax = plt.subplots(figsize=(15, 6))
bar_comparison(results, base_approaches=comparison_approaches, ax=ax)
plt.tight_layout()
fig.savefig(ROOT / "reports" / "f1-score-comparison-bar.png", bbox_inches="tight", dpi=150)
plt.show()
```
