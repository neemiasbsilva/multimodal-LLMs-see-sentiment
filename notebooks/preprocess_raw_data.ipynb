```python
def ia_calculation(sg):
    """
    Calcula a métrica IA (Image Agreement) com base no vetor de sentimento agrupado (sg).

    Parâmetros:
    sg (list ou tuple): Vetor de sentimento agrupado, onde cada elemento representa o número de votos para uma categoria de sentimento.

    Retorna:
    float: A métrica IA.
    """
    total_avaliadores = len(sg)
    max_votos = max(sg)
    IA = max_votos / total_avaliadores
    return IA

sentiment_data = {}

for id, caption in tqdm(zip(df["id"], df["caption"]), total=len(df), desc="Create supervised dataset"):
    sentiment_data[id] = {
        "sparse sentiment": [],
        "cluster sentiment": [],
        "perceptions": [],
        "caption": caption,
        "image_agreement": float,
    }

simple_sentiment = {
    "Positive": "Positive",
    "SlightlyPositive": "Positive",
    # "SlightlyPositive": "SlightlyPositive",
    "Neutral": "Neutral",
    "SlightlyNegative": "Negative",
    # "SlightlyNegative": "SlightlyNegative",
    "Negative": "Negative"
}
# sentiment_idx = {
#     "Positive": 4,
#     "SlightlyPositive": 3,
#     "Neutral": 2,
#     "SlightlyNegative": 1,
#     "Negative": 0
# }
sentiment_idx = {
    "Positive": 2,
    "Neutral": 0,
    "Negative": 1
}


for samples in tqdm(data["tasks"], desc="Image perceptions"):
    for sample in samples["images"]:
        id = sample["id"]
        sentiment = sample["sentiment"]
        # perceptions = ', '.join([str(per) for per in sample["perceptions"]])
        perceptions = [str(per) for per in sample["perceptions"]]
        
        if id in sentiment_data:
            # sentiment = [simple_sentiment[sent] for sent in sentiment]
            sentiment_data[id]["sparse sentiment"].append(sentiment)
            sentiment_data[id]["cluster sentiment"].append(simple_sentiment[sentiment])
            for perception in perceptions:
                sentiment_data[id]["perceptions"].append(perception)

for id in tqdm(sentiment_data, desc="Image agreement calculation"):
    sentiment = sentiment_data[id]["sparse sentiment"]
    sg = [0 for _ in range(len(sentiment))]
    counter = Counter(sentiment)
    for key in counter:
        sg[sentiment_idx[key]] = counter[key]
    sentiment_data[id]["image_agreement"] = ia_calculation(sg)
```

```python
def plot_sentiment_distribution(df):
    """"
    Plots the distribution of sentiments
    Parameters:
        df(dataframe): DataFrame containing the "sentiment" column.

    Returns:
        None
    """

    mapping = {0: "Negative", 1: "Positive"}
    # mapping = {1: "Negative", 0: "Neutral", 2: "Positive"}
    # mapping = {0: "Negative", 1: "SlightlyNegative", 2: "Neutral", 3: "SlightlyPositive", 4: "Positive"}
    df["sentiment"] = df["sentiment"].map(mapping)
    sentiment_counts = df["sentiment"].value_counts()

    plt.figure(figsize=(6, 4))
    sentiment_counts.plot(kind="bar", color="skyblue")
    plt.title("Sentiment Distribution")
    plt.xlabel("Sentiment")
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```

## Preprocess For Regression Approach

```python
import os
import shutil
import json
import pandas as pd
import numpy as np # Added for mean calculation
import nltk
from tqdm.notebook import tqdm
from collections import Counter
```

**Load Caption Data & Vote/Perception Data**

```python
df = pd.read_csv("../data/raw/image_caption_gpt4_openai.csv", delimiter=';')
display(df.head())
f = open("../data/raw/dataset.json")
data_json = json.load(f)
f.close()
```

**Processing Loop for each Approach (P5, P3, P2+, P2-)**

```python
for approach in range(0, 4):
    
    # 1. Define Mappings and Scale Values based on approach
    if approach == 0: ## P5 (Scale 0 to 4)
        p = 'p5'
        # Maps raw JSON strings to the approach categories
        simple_sentiment = {
            "Negative": "Negative",
            "SlightlyNegative": "SlightlyNegative",
            "Neutral": "Neutral",
            "SlightlyPositive": "SlightlyPositive",
            "Positive": "Positive"
        }
        # Maps categories to Regression Numbers
        sentiment_values = {
            "Negative": 0,
            "SlightlyNegative": 1,
            "Neutral": 2,
            "SlightlyPositive": 3,
            "Positive": 4,
        }

    elif approach == 1: ## P3 (Scale 0 to 2)
        p = 'p3'
        # Collapses Slightly into main categories
        simple_sentiment = {
            "Negative": "Negative",
            "SlightlyNegative": "Negative",
            "Neutral": "Neutral",
            "SlightlyPositive": "Positive",
            "Positive": "Positive"
        }
        sentiment_values = {
            "Negative": 0,
            "Neutral": 1,
            "Positive": 2,
        }

    elif approach == 2: ## P2+ (Scale 0 to 1)
        p = 'p2plus'
        # Neutral becomes Positive
        simple_sentiment = {
            "Negative": "Negative",
            "SlightlyNegative": "Negative",
            "Neutral": "Positive",
            "SlightlyPositive": "Positive",
            "Positive": "Positive"
        }
        sentiment_values = {
            "Negative": 0,
            "Positive": 1,
        }

    elif approach == 3: ## P2- (Scale 0 to 1)
        p = 'p2neg'
        # Neutral becomes Negative
        simple_sentiment = {
            "Negative": "Negative",
            "SlightlyNegative": "Negative",
            "Neutral": "Negative",
            "SlightlyPositive": "Positive",
            "Positive": "Positive"
        }
        sentiment_values = {
            "Negative": 0,
            "Positive": 1,
        }

    # 2. Initialize Data Structure
    sentiment_data = {}
    
    # Initialize entries for all captions in DF
    for id, caption in zip(df["id"], df["caption"]):
        sentiment_data[id] = {
            "numeric_votes": [], # We will store numbers here directly
            "caption": caption
        }

    # 3. Populate Votes from JSON
    # This matches the images in JSON to the IDs in our DataFrame
    for samples in tqdm(data_json["tasks"], desc=f"Processing Votes for {p}"):
        for sample in samples["images"]:
            id = sample["id"]
            raw_sentiment = sample["sentiment"]
            
            # Only process if this image ID exists in our caption dataframe
            if id in sentiment_data:
                # A. Map raw sentiment to approach sentiment (e.g., SlightlyPos -> Pos)
                cluster_s = simple_sentiment[raw_sentiment]
                
                # B. Map approach sentiment to Number (e.g., Pos -> 1)
                numeric_val = sentiment_values[cluster_s]
                
                # C. Append to list
                sentiment_data[id]["numeric_votes"].append(numeric_val)

    # 4. Calculate Regression Targets and Build Final Lists
    data_output = {
        "id": [],
        "text": [],
        "sentiment_score": [] # Continuous target
    }

    for id in tqdm(sentiment_data, desc=f"Calculating Regression {p}"):
        caption = sentiment_data[id]["caption"]
        votes = sentiment_data[id]["numeric_votes"]
        
        # Determine score
        if len(votes) > 0:
            # Calculate Mean of the votes
            final_score = np.mean(votes)
            
            data_output["id"].append(id)
            data_output["text"].append(caption)
            data_output["sentiment_score"].append(final_score)
        else:
            # Handle edge case where image might be in CSV but has no votes in JSON
            # Ideally this shouldn't happen if datasets align, but good practice to skip or flag
            pass 

    # 5. Save Dataframe
    output_filename = f"../data/gpt4-openai-regression/percept_dataset_regression_{p}.csv"
    pd.DataFrame(data_output).to_csv(output_filename, index=False)
    
    print(f"Saved {output_filename} with {len(data_output['id'])} samples.")
```
